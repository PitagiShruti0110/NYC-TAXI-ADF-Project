# Data Reading and Writing Creating Delta tables

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

Storage Variables

```python
silver = 'abfss://silver@nyctaxistorageshru.dfs.core.windows.net'
gold = 'abfss://gold@nyctaxistorageshru.dfs.core.windows.net'
```

DATA ZONE 

```python
df_zone =  spark.read.format('parquet')\
                    .option('inferSchema',True)\
                    .option('header',True)\
                    .load(f'{silver}/trip_zone')

df_zone.display()
```

```python
df_zone.write.format('delta')\
    .mode('append')\
    .option('path',f'{gold}/zone')\
    .saveAsTable('trip_zone')
```
