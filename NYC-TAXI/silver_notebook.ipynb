#Data Reading


**Importing Libraries

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

Reading CSV data

Trip Type Data

```python
df_trip_type = spark.read.format('csv')\
                    .option('inferSchema', True)\
                    .option('header', True)\
                    .load('abfss://bronze@nyctaxistorageshru.dfs.core.windows.net/trip_type')
```

```python
df_trip_type.display()
```

**Trip Zone**

```python
df_trip_zone = spark.read.format('csv')\
                    .option('inferSchema', True)\
                    .option('header', True)\
                    .load('abfss://bronze@nyctaxistorageshru.dfs.core.windows.net/trip_zone')
df_trip_zone.display()
```

**Trip Data**


```python
df_trip = spark.read.format('parquet')\
                    .schema(myschema)\
                    .option('header',True)\
                    .option('recursiveFileLookup',True)\
                    .load('abfss://bronze@nyctaxistorageshru.dfs.core.windows.net/trips2023data')  
```

```python
myschema = '''
    VendorID BIGINT,
    lpep_pickup_datetime TIMESTAMP,
    lpep_dropoff_datetime TIMESTAMP,
    store_and_fwd_flag STRING,
    RatecodeID BIGINT,
    PULocationID BIGINT,
    DOLocationID BIGINT,
    passenger_count BIGINT,
    trip_distance DOUBLE,
    fare_amount DOUBLE,
    extra DOUBLE,
    mta_tax DOUBLE,
    tip_amount DOUBLE,
    tolls_amount DOUBLE,
    ehail_fee DOUBLE,
    improvement_surcharge DOUBLE,
    total_amount DOUBLE,
    payment_type BIGINT,
    trip_type BIGINT,
    congestion_surcharge DOUBLE 
'''
```

```python
df_trip.display()
```

# Data Transformation

**Taxi Trip Type**

```python
df_trip_type = df_trip_type.withColumnRenamed('description','trip_description')
df_trip_type.display()
```

```python
df_trip_type.write.format('parquet')\
            .mode('append')\
            .option('path','abfss://silver@nyctaxistorageshru.dfs.core.windows.net/trip_type')\
            .save()
```

**Trip Zone**

```python

df_trip_zone = df_trip_zone.withColumn('zone1',expr("get(split(zone,'/'),0)"))\
                           .withColumn('zone2',expr("get(split(zone,'/'),1)"))

df_trip_zone.display()
```

```python
df_trip_zone.write.format('parquet')\
    .mode('append')\
    .option('path','abfss://silver@nyctaxistorageshru.dfs.core.windows.net/trip_zone')\
    .save()
```

**Trip Data**

```python
df_trip = df_trip.withColumn('trip_date',to_date('lpep_pickup_datetime'))\
                .withColumn('trip_year',year('lpep_pickup_datetime'))\
                .withColumn('trip_month',month('lpep_pickup_datetime'))
```

```python
df_trip.display()
```

```python
df_trip=df_trip.select('VendorID','PULocationID','DOLocationID','trip_distance','fare_amount','total_amount')
df_trip.display()
```

```python
df_trip.write.format('parquet')\
    .mode('append')\
    .option('path','abfss://silver@nyctaxistorageshru.dfs.core.windows.net/trips2023')\
    .save()
```
